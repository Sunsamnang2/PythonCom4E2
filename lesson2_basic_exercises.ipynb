## Exercise 1: Numeric analysis

1. Create one integer, one float in scientific notation, and one complex number.
2. Print each value and its type. For the complex number, also print its real and imaginary parts.
3. Convert the integer to a float and the float to an integer, then print the converted values.

```python
# Create an integer, a scientific-notation float, and a complex number
number_int = 12
number_float = 3.14e2
number_complex = 4 + 5j

print(number_int, type(number_int))
print(number_float, type(number_float))
print(number_complex, type(number_complex))
print("Real part:", number_complex.real)
print("Imaginary part:", number_complex.imag)

int_as_float = float(number_int)
float_as_int = int(number_float)
print("Integer as float:", int_as_float, type(int_as_float))
print("Float as integer:", float_as_int, type(float_as_int))
```

## Exercise 2: Safe casting challenge

Given `values = ["25", "3.14", "-8", "hello"]`, inspect each string with string methods and `if`/`elif`/`else`.
- Convert whole-number text to `int`.
- Convert decimal-number text to `float`.
- Keep anything else as a string.

Print each converted value and its type.

```python
values = ["25", "3.14", "-8", "hello"]
converted_values = []

for value in values:
    cleaned_value = value.strip()
    is_negative = cleaned_value.startswith("-")
    digits = cleaned_value[1:] if is_negative else cleaned_value

    if digits.isdigit():
        converted = int(cleaned_value)
    elif cleaned_value.count(".") == 1 and cleaned_value.replace(".", "", 1).lstrip("-").isdigit():
        converted = float(cleaned_value)
    else:
        converted = cleaned_value

    converted_values.append(converted)
    print(converted, type(converted))
```

## Exercise 3: String slicing matrix

Use `text = "DataSciencePython"` and print:
1. The first 4 characters.
2. The last 6 characters.
3. Every second character.
4. The text reversed.
5. `Science`, extracted using slicing only.

```python
text = "DataSciencePython"

print("First 4 characters:", text[:4])
print("Last 6 characters:", text[-6:])
print("Every second character:", text[::2])
print("Reversed:", text[::-1])
print("Science:", text[4:11])
```

## Exercise 4: String transformation pipeline

Start with `raw = " python, java, c++ , go"`.
1. Strip the outer spaces and split the text by comma.
2. Print the split list.
3. Loop through the list, strip each language, uppercase it, and print it.

```python
raw = " python, java, c++ , go"
languages = raw.strip().split(",")

print(languages)
for language in languages:
    print(language.strip().upper())
```

## Exercise 5: Formatting and escaping

1. Create variables `product`, `price`, and `discount`.

3. Print one line containing escaped quotes and another with newline and tab formatting.

```python
product = "notebook"
price = 12.50
discount = 0.20
discounted_price = price * (1 - discount)

print(f"Original price of {product}: ${price:.2f}")
print(f"Discounted price of {product}: ${discounted_price:.2f}")
print("She said, \"Python is fun!\"")
print(f"Product:\n\t{product}\n\tPrice: ${price:.2f}")
```
