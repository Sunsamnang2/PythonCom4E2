## Exercise 1: Tuple unpacking patterns

- Create a tuple with at least 7 values.
- Unpack into `first`, `second`, `*middle`, `last`.
- Print each variable and lengths where relevant.

```python
values = (10, 20, 30, 40, 50, 60, 70)
first, second, *middle, last = values
print("first:", first)
print("second:", second)
print("middle:", middle)
print("last:", last)
print("middle length:", len(middle))
```

## Exercise 2: Tuple transformation

- Start from a tuple of integers.
- Create a new tuple with each value multiplied by 10.
- Print original and transformed tuples.

```python
numbers = (1, 2, 3, 4, 5)
transformed = tuple(n * 10 for n in numbers)
print("Original:", numbers)
print("Transformed:", transformed)
```

## Exercise 3: Set-based analytics

- Given two sets of course names, print union, intersection, and both differences.
- Print symmetric difference and explain what it means in a comment.

```python
python_courses = {"Python", "Math", "Database", "AI"}
web_courses = {"Math", "AI", "Web", "Design"}
print("Union:", python_courses | web_courses)
print("Intersection:", python_courses & web_courses)
print("Python - Web:", python_courses - web_courses)
print("Web - Python:", web_courses - python_courses)
print("Symmetric difference:", python_courses ^ web_courses)
# Symmetric difference shows the courses that appear in exactly one of the two sets.
```

## Exercise 4: Set cleanup task

- Create a set from a list containing repeated and mixed-case words.
- Normalize words to lowercase before inserting into set.
- Print final normalized unique set sorted alphabetically.

```python
words = ["Apple", "banana", "apple", "Orange", "BANANA", "grape", "orange"]
normalized = {word.lower() for word in words}
print("Normalized unique set:", sorted(normalized))
```

## Exercise 5: Set from a list comprehension

- Use a list comprehension to build the squares of numbers 1..20 where the square is even.
- Convert the list to a set to remove duplicates (if any).
- Print the resulting set and its length.
- Convert it to a sorted list and print.

```python
squares = [n * n for n in range(1, 21) if (n * n) % 2 == 0]
unique_squares = set(squares)
print("Even squares:", unique_squares)
print("Length:", len(unique_squares))
print("Sorted list:", sorted(unique_squares))
```
